Q3. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the student enrollment. Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in "Course A" but not in "Course B" (difference). Find students who are only in one of the two courses (symmetric difference). course_a = {"John", "Alice", "Bob", "David"} course_b = {"Alice", "Eve", "Charlie", "David"}

Using set methods

• Find students who are in Course A but not in Course B (difference)

• Find students who are only in one of the two courses (symmetric difference)

Expected Output:

Students in Course A but not in Course B: {'Bob', 'John'}

Students only in one of the two courses: {'Eve', 'John', 'Bob', 'Charlie'}

```python

course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}

student_missing = course_a.difference(course_b)

only_student = course_a.symmetric_difference(course_b)

print(f"Students in Course A but not in Course B: {student_missing}")
print(f"Students only in one of the two courses: {only_student}")
```

Q4. numbers = (10, 20, 30, 40, 50, 60, 70)

Display the first 3 elements.
Display the last 3 elements.
Display alternate elements.

```python
numbers = (10, 20, 30, 40, 50, 60, 70)

# Display the first 3 elements.
first_elements = numbers[:3]

# Display the last 3 elements.
last_elements = numbers[-3:]

# Display alternate elements
alternate_elements = numbers[::2]

print(first_elements)
print(last_elements)
print(alternate_elements)
```

Q6. Set Comparison
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

Find common elements.
Find elements only in Set A.
Find elements only in Set B.

```python
A = {1, 2, 3, 4, 5} 
B = {4, 5, 6, 7}

# common elements.
common_elemnents = A.intersection(B)

#elements only in Set A
only_setA = A.difference(B)

#elements only in Set B
only_setB = B.difference(A)

print("Common elements:", common_elemnents)
print("Only in Set A:", only_setA)
print("Only in Set B:", only_setB)
```
