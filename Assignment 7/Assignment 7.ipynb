#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python

class Employee:

    # Constructor
    def __init__(self, name, number, address, salary, post,
                 joining_year, education, experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.experience = experience

    # Describe Employee
    def describe_employee(self):

        print("\n----- Employee Information -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Joining Year:", self.joining_year)
        print("Education:", self.education)
        print("Experience:", self.experience, "Years")

    # Update Experience
    def update_year_of_experience(self):

        self.experience += 1
        print("Updated Experience:", self.experience, "Years")

    # Increase Salary
    def increase_salary(self, amount):

        self.salary += amount
        print("Updated Salary:", self.salary)


# Creating Employee101

employee101 = Employee(
    "Moazzam", "101", "Karachi", 50000,
    "Web Developer", 2023, "BS Computer Science", 3
)


# Printing Four Attributes Individually

print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)


# Calling All Methods

employee101.describe_employee()

employee101.increase_salary(10000)

employee101.update_year_of_experience()


# Creating 9 More Employees

employee102 = Employee("Ali", "102", "Lahore", 60000, "PHP Developer", 2022, "BS IT", 4)

employee103 = Employee("Ahmed", "103", "Karachi", 55000, "Designer", 2021, "BS CS", 5)

employee104 = Employee("Hassan", "104", "Islamabad", 70000, "Frontend Developer", 2020, "BS CS", 6)

employee105 = Employee("Usman", "105", "Multan", 45000, "WordPress Developer", 2024, "DAE", 2)

employee106 = Employee("Bilal", "106", "Lahore", 80000, "Backend Developer", 2019, "BS CS", 7)

employee107 = Employee("Saad", "107", "Karachi", 65000, "UI Designer", 2022, "BS Design", 4)

employee108 = Employee("Hamza", "108", "Peshawar", 50000, "Python Developer", 2023, "BS IT", 3)

employee109 = Employee("Zain", "109", "Quetta", 90000, "Project Manager", 2018, "MBA", 8)

employee110 = Employee("Farhan", "110", "Karachi", 75000, "Laravel Developer", 2021, "BS CS", 5)


# Store All Employees in a List

employees = [
    employee101,
    employee102,
    employee103,
    employee104,
    employee105,
    employee106,
    employee107,
    employee108,
    employee109,
    employee110
]


# Display All 10 Employees

print("\n========== ALL EMPLOYEES ==========")

for employee in employees:
    employee.describe_employee()


# Salary Updates

print("\n========== SALARY UPDATES ==========")

employee101.increase_salary(10000)
employee102.increase_salary(5000)
employee103.increase_salary(8000)


# Experience Updates

print("\n========== EXPERIENCE UPDATES ==========")

employee101.update_year_of_experience()
employee102.update_year_of_experience()
employee103.update_year_of_experience()


# Display Updated Employee Information

print("\n========== UPDATED EMPLOYEES ==========")

employee101.describe_employee()
employee102.describe_employee()
employee103.describe_employee()
```
